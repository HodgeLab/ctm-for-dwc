```python
import logging
from pathlib import Path

from transportation_models.utils.data_processing import PeMSDataProcessor

logging.getLogger("matplotlib").setLevel(logging.WARNING)
logging.basicConfig(level=logging.INFO)
```

```python
# Parameters
detector_id = "400839"
save_dir = "/Users/robstallman/Downloads"  # Set to a Path to save plots, e.g. Path("./plots")
```

```python
processor = PeMSDataProcessor(
    root_directory="/Volumes/easystore/work/boulder/Caltrans/PeMS",
    save_transforms=False,
)

# Load and prepare data
df = processor.load_data_by_id(detector_id)
df = processor.standardize_timeseries(df)
df = df.loc[df["pct_observed"] >= processor.imputation_threshold].reset_index(drop=True)
df = processor.whiten_timeseries(df)
df = processor.filter_flow_outliers(df)

# Calibrate and plot
save_path = Path(save_dir) / f"{detector_id}.png" if save_dir else None
params = processor.extract_fundamental_diagram_params_from_timeseries(
    timeseries_df=df,
    detector_id=detector_id,
    make_plots=True,
    save_path=save_path,
)
params
```
